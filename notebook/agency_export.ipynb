```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\Junayed\nyc-311-resolution-time-analysis\data\clean\311_jan_week1_clean.csv")
```

```python
df.groupby("Agency")["Resolution_Days"].agg(
    Ticket_Count="count",
    Median="median",
    Mean="mean",
    P75=lambda x: x.quantile(0.75),
    P90=lambda x: x.quantile(0.90),
    Max='max'
).sort_values(by="Mean", ascending=False)
```

```python
# Filter for EDC, group by complaint type, and aggregate resolution days
EDC_df = df[df["Agency"] == "EDC"]

EDC_breakdown = (
   EDC_df.groupby("Complaint_Type")["Resolution_Days"]
    .agg(
        Ticket_Count="count",
        Median="median",
        Mean="mean",
        P75=lambda x: x.quantile(0.75),
        P90=lambda x: x.quantile(0.90),
        Max="max",
    )
    .sort_values(by="Ticket_Count", ascending=False)
)

# Add category percentage of EDC's total volume
EDC_breakdown["Pct_of_EDC"] = (EDC_breakdown["Ticket_Count"] / len(EDC_df) * 100).round(2)
EDC_breakdown["Cumulative_Pct"] = EDC_breakdown["Pct_of_EDC"].cumsum().round(2)
EDC_breakdown.head(10)
```

```python
# exporting EDC's data
EDC_df.to_csv("EDC_Tickets.csv", index=False)
EDC_breakdown.to_csv("EDC_Complaints.csv", index=False)
```
